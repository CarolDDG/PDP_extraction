```python
import matplotlib.pyplot as plt
import seaborn as sns


def plot_kde(
    dfs,
    metrics,
    figsize=None,
    fill=True,
    title_suffix="",
    dark=True,
):
    """
    Plot KDE distributions for multiple datasets.

    Parameters
    ----------
    dfs : dict
        {label: dataframe}

    metrics : list
        Metrics to plot

    figsize : tuple
        Figure size

    fill : bool
        Fill KDE area

    dark : bool
        Use dark background
    """

    if figsize is None:
        figsize = (5 * len(metrics), 4)


    # Theme
    if dark:
        plt.style.use("dark_background")
    else:
        plt.style.use("default")


    fig, axes = plt.subplots(
        1,
        len(metrics),
        figsize=figsize
    )

    if len(metrics) == 1:
        axes = [axes]


    for ax, metric in zip(axes, metrics):

        for label, df in dfs.items():

            sns.kdeplot(
                df[metric].dropna(),
                fill=fill,
                label=label,
                ax=ax
            )


        ax.set_title(
            metric.upper() + title_suffix,
            color="white" if dark else "black"
        )

        ax.set_xlabel(
            "Score",
            color="white" if dark else "black"
        )

        ax.set_ylabel(
            "Density",
            color="white" if dark else "black"
        )


        ax.legend()

        # Ax background
        if dark:
            ax.set_facecolor("black")

            ax.tick_params(
                colors="white"
            )

            ax.spines["bottom"].set_color("white")
            ax.spines["top"].set_color("white")
            ax.spines["left"].set_color("white")
            ax.spines["right"].set_color("white")


    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


def plot_bar_metrics(
    dfs,
    metrics,
    figsize=None,
    dark=True,
    title="Model Comparison",
    value_format="%.2f",
):
    """
    Plot bar chart comparing metrics across multiple datasets/models.
    """

    if figsize is None:
        figsize = (6, 4)


    if dark:
        plt.style.use("dark_background")
    else:
        plt.style.use("default")


    # Prepare data
    rows = []

    for label, df in dfs.items():
        for metric in metrics:
            rows.append({
                "Model": label,
                "Metric": metric.upper(),
                "Score": df[metric].mean()
            })

    plot_df = pd.DataFrame(rows)


    fig, ax = plt.subplots(figsize=figsize)


    sns.barplot(
        data=plot_df,
        x="Metric",
        y="Score",
        hue="Model",
        ax=ax
    )


    # Add values on top of bars
    for container in ax.containers:
        ax.bar_label(
            container,
            fmt=value_format,
            padding=3,
            color="white" if dark else "black",
            fontsize=10
        )


    ax.set_title(
        title,
        color="white" if dark else "black"
    )

    ax.set_xlabel("")

    ax.set_ylabel(
        "Score",
        color="white" if dark else "black"
    )


    if dark:
        ax.tick_params(colors="white")

        for spine in ax.spines.values():
            spine.set_color("white")


    ax.legend()

    plt.tight_layout()
    plt.show()
```

```python
def get_results(json_file):
    with open(json_file, "r", encoding="utf-8") as f:
        data = json.load(f)
    summary_df = pd.DataFrame([data["summary"]])
    items_df = pd.json_normalize(data["items"])
    return summary_df, items_df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


def plot_bar_metrics(dfs, metrics, title, figsize=(5,2)):
    """
    横向bar plot
    metrics: list，例如 ['precision', 'recall']
    """

    results = pd.DataFrame()

    for model in dfs.keys():
        for metric in metrics:
            results.loc[metric, model] = dfs[model][metric].mean()

    show = results.reset_index().melt(
        id_vars='index'
    )

    show.columns = ['Metric', 'Model', 'value']


    fig, ax = plt.subplots(figsize=figsize)

    sns.barplot(
        data=show,
        y='Metric',
        x='value',
        hue='Model',
        errorbar=None,
        palette="Set2",
        legend = False,
        ax=ax
    )

    # 数值
    for container in ax.containers:
        ax.bar_label(
            container,
            fmt="%.2f",
            padding=5,
            fontsize=10
        )

    # 半边框
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    ax.tick_params(
        axis='y',
        length=0
    )

    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.set_title(title)
    plt.yticks([])

    plt.tight_layout()
    plt.show()


# ======================
# 分组画图
# ======================

# 1. CER
plot_bar_metrics(
    dfs,
    ['cer'],
    'CER'
)


# 2. Precision + Recall
plot_bar_metrics(
    dfs,
    ['precision', 'recall'],
    'PRECISION & RECALL',
    figsize=(5,4)
)


# 3. Chunking Accuracy
plot_bar_metrics(
    dfs,
    ['chunking_accuracy'],
    'CHUNKING ACCURACY'
)


# 4. Merges + Splits
plot_bar_metrics(
    dfs,
    ['merges', 'splits'],
    'MERGES & SPLITS',
    figsize=(6,4)
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Model": [
        "PaddleOCR",
        "qwen-vl-max + qwen3.5-omni-flash",
        "Qwen3.5-0.8B",
        "GLM-OCR"
    ],
    "Speed": [16.34, 11.32, 6.77, 6.51],
    "Cost efficiency": [0.56, 5.94, 1.17, 0.46],
    "OCR Accuracy (F1)": [0.29, 0.82, 0.76, 0.77],
    "Chunking Accuracy": [0.44, 0.80, 0.70, 0.60]
})

df = df.loc[[2,3,1,0]]


data = df.copy()

# 越低越好的指标反向
for col in ["Speed", "Cost efficiency"]:
    data[col] = 1 - (
        (data[col] - data[col].min()) /
        (data[col].max() - data[col].min())
    )


categories = [
    "Speed",
    "Cost efficiency",
    "OCR Accuracy (F1)",
    "Chunking Accuracy"
]


N = len(categories)

angles = np.linspace(
    0,
    2*np.pi,
    N,
    endpoint=False
).tolist()

angles += angles[:1]


fig, ax = plt.subplots(
    figsize=(6,6),
    subplot_kw=dict(polar=True)
)


colors = sns.color_palette(
    "Set2",
    n_colors=len(data)
)


for (_, row), color in zip(data.iterrows(), colors):

    values = row[categories].tolist()
    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2.5,
        color=color,
        label=row["Model"]
    )

    ax.fill(
        angles,
        values,
        color=color,
        alpha=0.12
    )


ax.set_xticks(angles[:-1])
ax.set_xticklabels(
    categories,
    fontsize=12
)

ax.set_ylim(0,1)

ax.grid(
    color="gray",
    alpha=0.25,
    linewidth=0.8
)

ax.set_facecolor("#111111")

ax.legend(
    loc="center left",
    bbox_to_anchor=(1.5, 0.5),
    fontsize=12,
    frameon=False
)

plt.tight_layout()
plt.show()  

```
